6. Обучите модель линейной регрессии на рассматриваемых данных, написанную своими руками. Проиллюстрируйте работу модели графиком обучения и распределения целевой переменной. Выведите уравнение полученной гиперплоскости.

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X = pd.DataFrame(X)
cols = X.columns
X = scaler.fit_transform(X)
X = pd.DataFrame(data=X, columns=cols)
```

```python
X
```

```python
class Model:
    def __init__(self):
        self.b = None #инициализируем вектор коэффицциентов

    def predict(self, X):
        return X @ self.b #используем матричное умножение

    def MSE(self, X, Y):
        return (((Y - self.predict(X)).T @ (Y - self.predict(X))) / (2 * X.shape[0])).values[0][0]

    def fit(self, X, Y, alpha = 0.01, accuracy = 0.00001, max_steps = 10000, intercept=True):
        Y = np.array(Y).reshape(-1, 1)
        if intercept:
            X['intercept'] = 1 #добавляем столбец единиц для множественной линейной регрессии
        self.b = np.zeros((X.shape[1], 1))  #инициализация коэффициентов по подходящему размеру
        steps, errors = [], []
        step = 0
        old_err = self.MSE(X, Y) #будем считать точность ошибки

        for step in range(max_steps):
            dJ_b = X.T @ (self.predict(X) - Y) / X.shape[0]  #вычисляем градиент
            self.b -= alpha * dJ_b  #обновляем коэффициенты

            new_err = self.MSE(X, Y)
            step += 1
            if abs(new_err - old_err) < accuracy: #считаем точность ошибки, чтобы остановить, если изменения уже слишком малы
                break
            steps.append(step)
            errors.append(new_err)
            old_err = new_err

        return steps, errors
```

```python
m = Model()
steps, errors = m.fit(X, y)
J = m.MSE(X, y)
print('MSE после градиентного спуска: ', J)
```

```python
yy = m.predict(X)
plt.scatter(yy, y)
plt.plot(yy, yy, c='r')
```

По гистограмме видим, что модель оптимально предсказывает для данного датасета, однако из-за того, что значения имеют большое отклонение, средняя квадратичная ошибка также достаточно большая.

Распределение целевой переменной

```python
sns.histplot(y, bins=27)
```

```python
plt.subplot(1, 2, 2)
plt.plot(steps, errors, color='orange')
plt.title('Процесс обучения модели')
plt.xlabel('Шаги')
plt.ylabel('Ошибка')
plt.yscale('log')
plt.grid()
plt.show()
```
