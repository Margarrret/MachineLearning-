```python
from sklearn.datasets import load_breast_cancer
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = load_breast_cancer()
```

```python
X = data.data[:,:2]
y = data.target
```

```python
df = pd.DataFrame(data.data[:,:2], columns=data.feature_names[:2])
df['target'] = data.target
df['target'].value_counts()
```

```python
plt.figure(figsize=(15, 5))
for i in range(2):
    plt.subplot(1, 2, i+1)
    plt.hist(X[:, i])
    plt.xlabel(data.feature_names[i])
plt.suptitle('Гистограммы признаков',fontsize=14)
plt.show()
```

```python
from sklearn.naive_bayes import GaussianNB
gaussian_nb = GaussianNB()
```

```python
gaussian_nb.fit(X, y)
```

```python
y_pred = gaussian_nb.predict(X)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, f1_score
print(confusion_matrix(y, y_pred))
print('Accuracy= ', accuracy_score(y, y_pred))
print('F1_score= ', f1_score(y, y_pred))
```

На использованном в работе наборе данных примените другие вариации модели наивного Байеса - Мультономиальный, Бернулли, категориальные и комплементарный. Для каждой модели сделайте вывод о ее применимости.

```python
from sklearn.naive_bayes import MultinomialNB, BernoulliNB, ComplementNB, CategoricalNB

multinomial_nb = MultinomialNB()
multinomial_nb.fit(X, y)
y_pred_multinomial = multinomial_nb.predict(X)
print(f"MultinomialNB accuracy: {accuracy_score(y, y_pred_multinomial):.4f}")

bernoulli_nb = BernoulliNB()
bernoulli_nb.fit(X > X.mean(), y)
y_pred_bernoulli = bernoulli_nb.predict(X > X.mean())
print(f"BernoulliNB accuracy: {accuracy_score(y, y_pred_bernoulli):.4f}")

complement_nb = ComplementNB()
complement_nb.fit(X, y)
y_pred_complement = complement_nb.predict(X)
print(f"ComplementNB accuracy: {accuracy_score(y, y_pred_complement):.4f}")

categorial_nb = CategoricalNB()
categorial_nb.fit(X, y)
y_pred_categorial = categorial_nb.predict(X)
print(f"CategorialNB accuracy: {accuracy_score(y, y_pred_categorial):.4f}")
```
