```python
# количество пропущенных значений в исходной выборке, обозначенные как "NA"
n_NA = x_data.count('NA')
n_NA
```

```python
x = pd.DataFrame([eval(i) for i in x_data.split('; ') if i!='NA'])[0]
x
```

```python
# введите объем очищенной от пропусков выборки
n = len(x)
n
```

```python
# минимальное значение в вариационном ряду
x_min = x.min()
x_min
```

```python
# максимальное значение в вариационном ряду
x_max = x.max()
x_max
```

```python
# первую квартиль
Q1 = x.quantile(0.25)
Q1
```

```python
# медиану
Q2 = x.quantile(0.5)
Q2
```

```python
# третью квартиль
Q3 = x.quantile(0.75)
Q3
```

```python
# квартильный размах
RQ = Q3-Q1
RQ
```

```python
# среднее значение
x_sr = x.mean()
x_sr
```

```python
# стандартное отклонение (несмещенное)
s = x.std(ddof=1)
s
```

```python
# исправленную дисперсию
s2 = x.var(ddof=1)
s2
```

```python
# ошибку выборки
se = s/n ** 0.5
se
```

```python
# эксцесс
sts.kurtosis(x,bias=False)
```

```python
# коэффициент асимметрии
sts.skew(x, bias=False)
```

```python
# левую границу 0.99-доверительного интервала для E(X)
gamma=0.99
n = len(x)
xsr=np.mean(x)
s2 = sum((x-xsr)**2/(n-1))
s = math.sqrt(s2)
sts.t.interval(gamma, n-1, xsr, s/math.sqrt(n))[0]
```

```python
# правую границу 0.99-доверительного интервала для E(X)
sts.t.interval(gamma, n-1, xsr, s/math.sqrt(n))[1]
```

```python
# левую границу 0.99-доверительного интервала для Var(X)
s2 = sum((x-x_sr)**2)/(n-1)
# s2=np.var(x,ddof=1)
# s2
chi2_gamma1 = sts.chi2.ppf((1-gamma)/2,n-1)
print((n-1)*s2/chi2_gamma1)
```

```python
# правую границу 0.99-доверительного интервала для Var(X)
chi2_gamma2 = sts.chi2.ppf((1+gamma)/2,n-1)
print((n-1)*s2/chi2_gamma2)
```
